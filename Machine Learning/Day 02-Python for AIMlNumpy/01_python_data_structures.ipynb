## Task 1
Create a list of 5 student marks and print:
1. Full list
2. First mark
3. Last mark
4. Average marks

````python
stu=[10,20,30,40,50]
print("Full list:",stu)
print("First element:",stu[0])
print("Last element:",stu[-1])
avg=sum(stu)/len(stu)
print("Average:",avg)
````

## Task 4
Create a list of labels:
```
["cat", "dog", "cat", "bird", "dog"]

Convert it into a set and print unique labels.


````python
animal=["cat", "dog", "cat", "bird", "dog"]
unique_animals=set(animal)
print("Unique animals:", unique_animals)
````

## Task 5
Create this dataset:
```
dataset = [    [10, 20, 30],    [40, 50, 60],    [70, 80, 90]]
```
Print:
```
First row
Second row
Value 50
Value 90

````python
dataset=[[10,20,30],[40,50,60],[70,80,90]]
print("Dataset:", dataset)
for row in dataset:
     print("Row:", row)

for row in dataset:
    for value in row:
        print("Value:", value)
````
