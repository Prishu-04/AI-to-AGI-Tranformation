### Basic Error Example

```python
num1 = 10
num2 = 0
result = num1 / num2
print(result)
```

### Try-execept Block

### Syntax

```python
try:
    risky_code
except:
    error_handling_code
```

### Example

```python
try:
    num1 = 10
    num2 = 0
    result = num1 / num2
    print(result)
except:
    print("Something went wrong")
```

## Multiple Error handing

```python
try:
    marks = [80, 90, 70]
    print(marks[5])
except IndexError:
    print("Index does not exist")
except ZeroDivisionError:
    print("Cannot divide by zero")
except:
    print("Unknown error occurred")
```

### Try-except-else


```python
try:
    num1 = 20
    num2 = 5
    result = num1 / num2
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Result:", result)
```

## Finally


```python
try:
    num1 = 20
    num2 = 0
    result = num1 / num2
except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("Program completed")
```

## AI/ML Use Case of error handling

```python
try:
    accuracy = 92 / 100
    print("Model Accuracy:", accuracy)
except:
    print("Error while calculating model accuracy")
```

## 9. Writing to a File

```python
file = open("student_report.txt", "w")
file.write("Student Name: Pratyaksh\n")
file.write("Average Marks: 85\n")
file.write("Result: Pass\n")
file.close()
```

## 10. Reading of File

```python
file = open("student_report.txt", "r")
content = file.read()
print(content)
file.close()
```

## 11. Better Method : with open()

```python
with open("student_report.txt", "w") as file:
    file.write("Student Name: Pratyaksh\n")
    file.write("Average Marks: 85\n")
    file.write("Result: Pass\n")
```

#### Reading 

```python
with open("student_report.txt", "r") as file:
    content = file.read()
    print(content)
```
