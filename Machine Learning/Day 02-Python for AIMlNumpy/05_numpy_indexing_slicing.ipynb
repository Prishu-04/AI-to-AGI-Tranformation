## Task 1
Create this array:
```
arr = np.array([10, 20, 30, 40, 50, 60])
```
Print:
```
First element.
Last element
Elements from index 1 to 4
Every second element
```

````python
arr=np.array([10,20,30,40,50,60])
print("First element:",arr[0])
print("Last element:",arr[-1])
print("Elements from index 1 to 4:",arr[1:5])
print("Every 2nd element:",arr[0::2])
````

## Task 2
Create this 2D dataset:
```
dataset = np.array([    
	[85, 90, 78],
    [70, 88, 92],
    [60, 75, 80]
])
```
Print:
```
First row
Second row
Value 88
Value 80
First column
Second column

````python
dataset=np.array([
     [85,90,78],
     [70,88,92],
     [65,75,80]
])
print("Dataset:\n",dataset)
print("First row",dataset[0])
print("Second row",dataset[1])
print("value 88",dataset[1,1])
print("Value 80",dataset[2,2])
print("First coumn",dataset[:,0])
print("Second column",dataset[:,1])
````

## Task 4
Create array from 1 to 12.
Convert it into:
```
3 rows and 4 columns
2 rows and 6 columns


````python
arr=np.arange(1,13)
reshape_arr1=arr.reshape(3,4)
reshape_arr2=arr.reshape(2,6)
print("Original array:",arr)
print("Reshaped array (3,4):\n",reshape_arr1)
print("Reshaped array (2,6):\n",reshape_arr2)
````
