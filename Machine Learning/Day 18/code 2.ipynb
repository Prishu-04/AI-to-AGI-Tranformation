```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_iris
iris=load_iris(as_frame=True)
x=iris.data
y=iris.target
print("Features data:",x.head())
```

```python
y=iris.target
print("Target:",y.head())
```

```python
print("Feature Names:",iris.feature_names)
print("\nClass Names:",iris.target_names)
```

```python
print("Shape:",x.shape)
```

```python
print("\nClass Distribution:",y.value_counts().sort_index())
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(
     x,y,test_size=0.20,random_state=42,stratify=y
)
print("X_train shape:",X_train.shape)
print("X_test shape:",X_test.shape)
print("\nTraining Classes:",y_train.value_counts().sort_index())
print("\nTesting Classes:",y_test.value_counts().sort_index())
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
knn_pipeline=Pipeline(steps=[
     ("scaler",StandardScaler()),
     ("classifier",KNeighborsClassifier(
          n_neighbors=5,
          weights="uniform"
     ))
])
```

```python
knn_pipeline.fit(X_train,y_train)
```

```python
y_pred=knn_pipeline.predict(X_test)
print("Actual Classes")
print(y_test.to_numpy())
print("\nPredicted Classes:")
print(y_pred)
```

```python
actual_names=iris.target_names[y_test.to_numpy()]
predicted_names=iris.target_names[y_pred]
results=X_test.copy()
results["Actual"]=actual_names
results["Predicted"]=predicted_names
results.head()
```

```python
from sklearn.metrics import accuracy_score
acc=accuracy_score(y_test,y_pred)
print("Accuracy:",acc)
```

```python
probab=knn_pipeline.predict_proba(X_test)
probab_df=pd.DataFrame(    
     probab,    
     columns=iris.target_names,    
     index=X_test.index
)
print(probab_df.head())
```

```python
from sklearn.model_selection import cross_val_score
k_val=range(1,21)
cv_results=[]
for k in k_val:
     can_model=Pipeline(steps=[
          ("scaler",StandardScaler()),
          ("classifier",KNeighborsClassifier(
               n_neighbors=k,
          ))
     ])
     scores=cross_val_score(
          can_model,
          X_train,
          y_train,
          cv=5,
          scoring="accuracy"
     )
     cv_results.append({
          "K":k,
          "Mean_CV_Accuracy":scores.mean(),
          "CV_std":scores.std()
     })     
k_results_df=pd.DataFrame(cv_results)
print(k_results_df)
best_row = k_results_df.sort_values(
     ["Mean_CV_Accuracy", "CV_std"],ascending=[False, True]
).iloc[0]
best_k = int(best_row["K"])
print("\nBest K:", best_k)
     
```

```python
best_knn = Pipeline(steps=[
    ("scaler", StandardScaler()),
    (
        "classifier",
        KNeighborsClassifier(
            n_neighbors=best_k
        )
    )
])
best_knn.fit(X_train, y_train)
final_predictions = best_knn.predict(X_test)
final_accuracy = accuracy_score(y_test, final_predictions)
print("Final Test Accuracy:", final_accuracy)
```

```python
weight_results = []
for weight_method in ["uniform", "distance"]:
    model = Pipeline(steps=[
        ("scaler", StandardScaler()),
        (
            "classifier",
            KNeighborsClassifier(
                n_neighbors=best_k,
                weights=weight_method
            )
        )
    ])
    scores = cross_val_score(
        model,
        X_train,
        y_train,
        cv=5,
        scoring="accuracy"
    )
    weight_results.append({
        "Weights": weight_method,
        "Mean_CV_Accuracy": scores.mean()
    })
print(pd.DataFrame(weight_results))
```

```python
scaler = knn_pipeline.named_steps["scaler"]
classifier = knn_pipeline.named_steps["classifier"]
```

```python
sample = X_test.iloc[[0]]
sample_scaled = scaler.transform(sample)
```

```python
distances, indices = classifier.kneighbors(sample_scaled)
print("Distances:")
print(distances)
print("\nNeighbor Positions in Training Array:")
print(indices)
```

```python
neighbor_rows = X_train.iloc[indices[0]].copy()
neighbor_rows["Class"] = iris.target_names[
     y_train.iloc[indices[0]].to_numpy()]
neighbor_rows["Distance"] = distances[0]
neighbor_rows
```
