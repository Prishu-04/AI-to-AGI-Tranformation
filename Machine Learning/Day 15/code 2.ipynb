## Classification Dataset

```python
classification_data = [
    {"cgpa": 8.5, "projects": 4, "placed": "Yes"},
    {"cgpa": 6.2, "projects": 1, "placed": "No"},
    {"cgpa": 7.9, "projects": 3, "placed": "Yes"},
    {"cgpa": 5.8, "projects": 0, "placed": "No"}
]

for row in classification_data:
    print(row['placed'])
```

## Regression Dataset Example

```python
regression_data = [
    {"study_hours": 8, "attendance": 90, "marks": 88},
    {"study_hours": 2, "attendance": 45, "marks": 40},
    {"study_hours": 6, "attendance": 75, "marks": 72},
    {"study_hours": 4, "attendance": 60, "marks": 55}
]

for row in regression_data:
    print(row)
```

### Clustering Dataset Example

```python
clustering_data = [
    {"age": 21, "monthly_spend": 3000},
    {"age": 22, "monthly_spend": 3500},
    {"age": 45, "monthly_spend": 12000},
    {"age": 47, "monthly_spend": 13000}
]

for row in clustering_data:
    print(row)
```
