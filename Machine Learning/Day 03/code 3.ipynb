# Vectorized Operations and Broadcasting

```python
import numpy as np
```

without Numpy

```python
marks=[85,72,45,91,60]
sc_marks=[]
for mark in marks:
     sc_marks.append(mark/100)

print(sc_marks)
```

With numpy

```python
marks=np.array(marks)
sc_marks=marks/100
print(sc_marks)
```

## Element Wise Arithmetic

```python
import numpy as np
marks = np.array([85, 72, 45, 91, 60])
print(marks + 5)
print(marks - 10)
print(marks * 2)
print(marks / 100)
```

## Vectorised Operation on 2d Dataset

```python
students = np.array([
    [85, 90, 5],
    [72, 80, 4],
    [45, 70, 3],
    [91, 95, 6],
    [60, 78, 2]
])
print(students)
```

```python
sc_marks=students/100
print(sc_marks)
```

## Column wise scaling

```python
marks = students[:, 0]
attendance = students[:, 1]
study_hours = students[:, 2]

scaled_marks = marks / 100
scaled_attendance = attendance / 100
scaled_study_hours = study_hours / 10

print("Scaled Marks:", scaled_marks)
print("Scaled Attendance:", scaled_attendance)
print("Scaled Study Hours:", scaled_study_hours)
```

## Broadcasting

```python
marks=np.array([85,72,45,91,60])
bonus=marks+5
print(bonus)
```

## Broadcasting with 2D array with 1D arrays

```python
students = np.array([
    [85, 90, 5],
    [72, 80, 4],
    [45, 70, 3],
    [91, 95, 6],
    [60, 78, 2]
])
scale_values = np.array([100, 100, 10])
scaled_students = students / scale_values
print(scaled_students)
```

```python
print(students.shape)
print(scale_values.shape)
```

## Add weights to features.

```python
wgt=np.array([0.5,0.3,0.2])
weighted_students=scaled_students*wgt
print(weighted_students)
```

## Calculate final score

```python
final_scores = np.sum(weighted_students, axis=1)
print(final_scores)
```

```python
excellent = final_scores >= 0.75
print(excellent)
```
