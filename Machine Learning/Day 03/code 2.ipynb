# 05_numpy_indexing_slicing

```python
import numpy as np
```

### Indexing

```python
arr=np.array([1,2,3,4,5])
for i in range(arr.size):
    print("At index",i,":",arr[i])
```

### Negative Indexing

```python
arr=np.array([10,20,30,40,50])
for i in range((-arr.size),0):
    print("At index",i,":",arr[i])
```

### Slicing in NumPy

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[1:4])
```

## Common Slicing Example

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[:3])
print(arr[2:])
print(arr[:])
print(arr[::2])
```

## Indexing in 2D Array

```python
dataset = np.array([
    [85, 90, 78],
    [70, 88, 92],
    [60, 75, 80]
])
```

### Accessing the value

```python
for i in range(dataset.shape[0]):
    for j in range(dataset.shape[1]):
        print("At row",i,"and column",j,":",dataset[i][j])
```

## Slicing Rows in 2D Array

```python
for i in range(dataset.shape[0]):
     print("Row",i,":",dataset[i])
```

## Slicing Columns in 2D Array

```python
for i in range(dataset.shape[0]):
     print("Column",i,":",dataset[:,i])
```

## Slicing Multiple ROws and Columns

```python
for i in range(dataset.shape[0]):
    for j in range(dataset.shape[1]):
        print(dataset[0:i,0:j])
```

## Use case of AI/ML

```python
data = np.array([
    [85, 90, 78, 1],
    [70, 88, 92, 1],
    [35, 40, 45, 0],
    [95, 92, 88, 1]
])

X = data[:, 0:3]
y = data[:, 3]

print("Features:")
print(X)

print("Labels:")
print(y)
```

## Reshape

```python
arr = np.array([1, 2, 3, 4, 5, 6])
new_arr = arr.reshape(2, 3)
print(new_arr)
```

### More Examples of Reshape

```python
arr = np.arange(1, 13)
arr_2d = arr.reshape(3, 4)
arr_3d = arr.reshape(2, 2, 3)
print(arr_2d)
print(arr_3d)
```

### Flattening Array

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
flat = arr.flatten()
print(flat)
```

## Common Errors and Corrections

### Error 1: Index out of index

```python
arr = np.array([10, 20, 30])
print(arr[5])
```

### Error 2: Wrong 2D Index

```python
dataset = np.array([
    [85, 90, 78],
    [70, 88, 92]
])
print(dataset[2, 1])
```

### Error 3: Invalid Reshape

```python
arr = np.array([1, 2, 3, 4, 5, 6])
new_arr = arr.reshape(4, 2)
```

## Boolean Filtering

```python
students = np.array([[100, 85, 90], [75, 80, 82], [60, 70, 65]])
marks=students[:,0]
mask=marks>=80
print(mask)
```

## Filtering multiple Conditions

```python
marks = students[:, 0]
attendance = students[:, 1]
mask = (marks >= 80) & (attendance >= 85)
selected_students = students[mask]
print(selected_students)
```
