## Difference Between Python List and NumPy Array

### Python List

```python
marks=[10,20,30]
print(marks*2)
```

### NumPy Array

```python
marks=np.array([10,20,30])
print(marks*2)
```

## Creation- of 2D Array

## Checking Arrays Dimensions

```python
arr1=np.array([1,2,3])
arr2=np.array([[4,5,6],[7,8,9]])
print(arr1.ndim)
print(arr2.ndim)
```

### Checking Shape

```python
dataset = np.array([
    [85, 90, 78],
    [70, 88, 92],
    [60, 75, 80]
])

print(dataset.shape)
```

### Checking Size

```python
dataset = np.array([
    [85, 90, 78],
    [70, 88, 92],
    [60, 75, 80]
])

print(dataset.size)
```

## Checking Data Type]

```python
arr = np.array([10, 20, 30])
print(arr.dtype)
```

## Creating Special Arrays

### Array of Zeros

```python
zero=np.zeros(5)
print(zero)
```

### Arrays of ones

```python
one=np.ones((2,3))
print(one)
```

### 2d Zeros Array

```python
zero_2d=np.zeros((3,4))
print(zero_2d)
```

### Range Array

```python
arr=np.arange(1,10)
print(arr)
```

### Range With Step

```python
arr2=np.arange(0,20,3)
print(arr2)
```

### Evenly spaced Values

```python
value=np.linspace(0,1,5)
print(value)
```

## Basic Mathematical Operation

```python
arr3 = np.array([10, 20, 30])

print(arr3 + 5)
print(arr3 - 5)
print(arr3 * 2)
print(arr3 / 2)
```
