## Task 1
Create a NumPy array:
```
[10, 20, 30, 40, 50]
```
Print:
```
Array
Dimension
Shape
Size
Data type


````python
arr=np.array([10,20,30,40,50])
print("Array",arr)
print("Dimension",arr.ndim)
print("Shape",arr.shape)
print("Size",arr.size)
print("Data type",arr.dtype)
````

[    [85, 90, 78],    [70, 88, 92],    [60, 75, 80]]
