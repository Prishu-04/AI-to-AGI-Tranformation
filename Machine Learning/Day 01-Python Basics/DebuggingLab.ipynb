# Debugging Lab

### 1. How to Read a Python Error

```python
marks = input("Enter marks: ")
total = marks + 10
print(total)
```

### Bug 1: SyntaxError

```python
marks = 80

if marks >= 40
    print("Pass")
```

### Bug 2: Indentation Error

```python
marks = 80

if marks >= 40:
print("Pass")
```

### Bug 3 : Type Error

```python
marks = input("Enter marks: ")
bonus = 5
final_marks = marks + bonus
print(final_marks)
```

### Bug 4: ValueError

```python
marks = int(input("Enter marks: "))
print(marks)
```

### Bug 5 : Index Error

```python
students = []
topper = students[0]
print(topper)
```

### Bug 6 : Key Error

```python
student = {
    "name": "Rahul",
    "marks": 85
}

print(student["grade"])
```

### Zero Divisor Error

```python
students = []
average = 100 / len(students)
print(average)
```
