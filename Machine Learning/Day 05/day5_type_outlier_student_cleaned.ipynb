```python
import pandas as pd
import numpy as np
```

## Messy Dataset

```python
data = {
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "name": ["Rahul", "Aman", "Priya", "Sneha", "Karan", "Anjali", "Rohit", "Neha", "Vikas", "Riya"],
    "marks": [85, "72", "sixty", 91, 120, 88, 35, -10, "95", 67],
    "attendance": [90, "80", 70, 95, 85, 150, -20, 75, "ninety", 82],
    "study_hours": [5, "4", 3, 6, 5, 50, 1, "two", 7, -3],
    "subject": ["AI", "Data", "ai", "AI", "Ai", "DATA", "Web", "web", "data", "WEB"]
}
```

```python
df=pd.DataFrame(data)
df
```

## Convert Columns to numeric

```python
df["marks"]=pd.to_numeric(df["marks"],errors="coerce")
df["attendance"]=pd.to_numeric(df["attendance"],errors="coerce")
df["study_hours"]=pd.to_numeric(df["study_hours"],errors="coerce")
df
```

## Filling missing values

```python
df["marks"]=df["marks"].fillna(df["marks"].median())
df["attendance"]=df["attendance"].fillna(df["attendance"].median())
df["stuyd_hours"]=df["study_hours"].fillna(df["study_hours"].median())
df
```

## handle invalid range values

### Remove Invalid rows:

```python
df=df[(df["marks"]>=0)& (df["marks"]<=100)]
df=df[(df["attendance"]>=0)& (df["attendance"]<=100)]
df=df[(df["study_hours"]>=0)& (df["study_hours"]<=100)]
df
```

### Clip Outliers

```python
df["marks"] = df["marks"].clip(lower=0, upper=100)
df["attendance"] = df["attendance"].clip(lower=0, upper=100)
df["study_hours"] = df["study_hours"].clip(lower=0, upper=12)
df
```

## Convert subject to uppercase

```python
df["subject"]=df["subject"].str.upper()
df
```

```python
df.to_csv("day5_type_outlier_student_cleaned.csv",index=False)
```
