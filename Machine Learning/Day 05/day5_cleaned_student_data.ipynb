```python
import pandas as pd
import numpy as np
```

## Messy Student Dataset

```python
data = {
    "student_id": [101, 102, 103, 104, 105, 106, 106, 108, 109, 110],
    "name": ["Rahul", "Aman", "Priya", "Sneha", None, "Anjali", "Anjali", "Rohit", "Karan", "Neha"],
    "marks": [85, 72, None, 91, 120, 88, 88, 35, "sixty", 45],
    "attendance": [90, 80, 70, 95, 85, 92, 92, -10, 78, None],
    "study_hours": [5, 4, 3, 6, 5, 5, 5, 1, 2, None],
    "subject": ["AI", "Data", "ai", "AI", "Ai", "DATA", "DATA", "Web", "web", "Data"]
}
```

## Creation of DataFrame

```python
df=pd.DataFrame(data)
df
```

## Check Missing Values

```python
print(df.isnull().sum())
```

## Check Duplicate value

```python
print(df.duplicated().sum())
```

## Remove the duplicate value

```python
print(df.drop_duplicates())
df
```

## Handling missing values

### Option A Drop rows with missing values

```python
df_drop=df.dropna()
df_drop
```

### Option B Fill missing values

```python
df["name"]=df["name"].fillna("Unknown")
df["attendance"]=df["attendance"].fillna(df["attendance"].mean()).round(2)
df["study_hours"]=df["study_hours"].fillna(df["study_hours"].mean()).round(2)

df
```

## Fix invalid values

```python
df=df[df["attendance"]>=0]
df
```

## Fix Wrong datatype

```python
df["marks"]=pd.to_numeric(df["marks"],errors="coerce")
df
```

```python
df["marks"]=df["marks"].fillna(df["marks"].median())
df
```

## Fix data inconsistency

```python
df["subject"]=df["subject"].str.upper()
df
```

## save to csv file

```python
df.to_csv("day5_cleaned_student_data.csv",index=False)
```
