```python
import numpy as np
import pandas as pd
```

## Messy dataset

```python
data = {
    "student_id": [101, 102, 103, 104, 105, 106, 107],
    "name": ["Rahul", "Aman", None, "Sneha", "Karan", "Anjali", "Rohit"],
    "marks": [85, None, 45, 91, np.nan, 88, 35],
    "attendance": [90, 80, None, 95, 78, np.nan, 60],
    "study_hours": [5, 4, 3, None, 2, 5, np.nan],
    "subject": ["AI", "Data", "Web", "AI", None, "Data", "Web"]
}
```

```python
df=pd.DataFrame(data)
df
```

## Check NULL Values

```python
print(df.isnull())   
```

```python
print(df.isnull().sum())
```

## 4. `dropna()` - remove missing value

```python
drop_df=df.dropna()
print(drop_df)
```

## 5. Drop rows only when specific columns are missing

```python
df_drop=df.dropna(subset=["name","marks"])
print(df_drop)
```

## `fillna()`- fill missing data

### Fill text columns

```python
df["name"]=df["name"].fillna("Unknown")
df["subject"]=df["subject"].fillna("Unknown")
df
```

### Fill numeric columns

```python
df["marks"] = df["marks"].fillna(df["marks"].median())
df["attendance"] = df["attendance"].fillna(df["attendance"].mean())
df["study_hours"] = df["study_hours"].fillna(df["study_hours"].median())
df
```

```python
df.to_csv("day5_null_handled_student_data.csv",index=False)
```
