```python
import pandas as pd
import numpy as np
```

## Messy Dataset

```python
data={
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Rahul", "Aman", "Priya", "Sneha", "Karan", "Anjali", "Rohit", "Neha"],
    "marks": [85, "72", "sixty", 91, 120, 88, 35, -10],
    "attendance": [90, "80", 70, 95, 85, 150, -20, 75],
    "study_hours": [5, "4", 3, 6, 5, 50, 1, "two"],
    "subject": ["AI", "Data", "ai", "AI", "Ai", "DATA", "Web", "web"]
}
```

```python
df=pd.DataFrame(data)
df
```

## COnvert the columns to numeric:

```python
df["marks"]=pd.to_numeric(df["marks"],errors="coerce")
df["attendance"]=pd.to_numeric(df["attendance"],errors="coerce")
df["study_hours"]=pd.to_numeric(df["study_hours"],errors="coerce")
df
```

## Fill missing values after conversion

```python
df["marks"] = df["marks"].fillna(df["marks"].median())
df["study_hours"] = df["study_hours"].fillna(df["study_hours"].median())
df
```

## Handle Invalide Range Values

### Method 1: Remove invalid rows

```python
df = df[(df["marks"] >= 0) & (df["marks"] <= 100)]
df = df[(df["attendance"] >= 0) & (df["attendance"] <= 100)]
df = df[(df["study_hours"] >= 0) & (df["study_hours"] <= 12)]
df
```

### Method 2: Clip Outliers

```python
df["marks"] = df["marks"].clip(lower=0, upper=100)
df["attendance"] = df["attendance"].clip(lower=0, upper=100)
df["study_hours"] = df["study_hours"].clip(lower=0, upper=12)
df
```

## 7. Standardize text column

```python
df["subject"] = df["subject"].str.upper()
df
```

## IQR Method for Outliers

```python
q1 = df["marks"].quantile(0.25)
q3 = df["marks"].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

df_without_outliers = df[
    (df["marks"] >= lower_bound) &
    (df["marks"] <= upper_bound)
]

print(df_without_outliers)
```
