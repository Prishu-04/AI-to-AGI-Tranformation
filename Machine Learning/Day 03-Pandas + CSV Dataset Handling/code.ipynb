## Pandas Series

```python
import pandas as pd
marks = pd.Series([85, 90, 78, 92])
print(marks)
```

## Series with Custom Index

```python
import pandas as pd
marks = pd.Series(
    [85, 90, 78],
    index=["Math", "Science", "English"]
)
print(marks)
```

### Accessing Series Vales

```python
import pandas as pd
marks = pd.Series(
    [85, 90, 78],
    index=["Math", "Science", "English"]
)
for i in range(len(marks)):
     print(f"{marks.index[i]}: {marks.iloc[i]}")
```

```python
print(marks["Math"])
print(marks["English"])
print(marks["Science"])
```

## Pandas Data frame

```python
import pandas as pd
data = {
    "Name": ["Pratyaksh", "Rahul", "Sneha"],
    "Math": [85, 70, 90],
    "Science": [88, 75, 95],
    "English": [78, 80, 92]
}
df = pd.DataFrame(data)
print(df)
```

## Checking Data frame Columns:

```python
import pandas as pd
data = {
    "Name": ["Pratyaksh", "Rahul", "Sneha"],
    "Math": [85, 70, 90],
    "Science": [88, 75, 95],
    "English": [78, 80, 92]
}
df = pd.DataFrame(data)
print(df.columns)
```

## Checking Data Frame Shape:

```python
print(df.shape)
```

## Checking First Rows using `head()`

```python
print(df.head(2))
```

## Checking Last Row using `tail()`

```python
print(df.tail(1))
```

## Selecting One Column

```python
print(df["Name"])
```

## Selecting Multiple Columns

```python
print(df[["Name", "Math"]])
```

## Creating New Column

```python
df["Total"] = df["Math"] + df["Science"] + df["English"]
print(df)
```

## 17. Creating Average Column

```python
df["Average"] = df["Total"] / 3
df
```

To round off avergaes:

```python
df["Average"]= df["Average"].round(2)
df
```

## 18. Adding Result Column

```python
df["Result"]=["Pass" if avg >= 50 else "Fail" for avg in df["Average"]]
df
```
