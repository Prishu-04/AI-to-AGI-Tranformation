## Filtering Data

```python
high_math = df[df["Math"] > 80]
print(high_math)
```

## Filter Students with Attendance Above 90

```python
high_attendance = df[df["Attendance"] > 90]
print(high_attendance)
```

## Filter Students with Study Hours More Than 4

```python
hardworking_students = df[df["Study_Hours"] > 4]
print(hardworking_students)
```

## 13. Multiple Conditions

```python
result = df[(df["Math"] > 80) & (df["Attendance"] > 90)]
print(result)
```

```python
result = df[(df["Math"] > 90) | (df["Science"] > 90)]
print(result)
```

## Creating Total and Average Columns

```python
df["Total"] = df["Math"] + df["Science"] + df["English"]
df["Average"] = df["Total"] / 3
df["Average"] = df["Average"].round(2)
df
```

## Filter Passed Students
Pass condition:

Average >= 50

```python
passed_students = df[df["Average"] >= 50]
print(passed_students)
```

```python

16. Filter Failed Students
failed_students = df[df["Average"] < 50]

print(failed_students)
```

## Add Result Column

```python
df["Result"] = ["Pass" if avg >= 50 else "Fail" for avg in df["Average"]]
df
```

## Add Grade

```python
def assign_grade(avg):
    if avg >= 90:
        return "A+"
    elif avg >= 75:
        return "A"
    elif avg >= 60:
        return "B"
    elif avg >= 50:
        return "C"
    else:
        return "Fail"
df["Grade"] = df["Average"].apply(assign_grade)
df
```
