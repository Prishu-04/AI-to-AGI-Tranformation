Task 5
Create:
```
Total column
Average column
Result column
Grade column
```
Then separate:
```
X = Math, Science, English, Attendance, Study_Hoursy = Result
```
---


````python
df["Total_Score"] = df["Math"] + df["Science"] + df["English"]
````

````python
df["Average_Score"] = df["Total_Score"] / 3
df["Avg"]=df["Average_Score"].round(2)
````

````python
df["Result"]=["Pass" if x >= 60 else "Fail" for x in df["Avg"]]
````

````python
def assign_grades(avg):
     if avg >= 90:
          return "A"
     elif avg >= 80:
          return "B"
     elif avg >= 70:
          return "C"
     elif avg >= 60:
          return "D"
     else:
          return "Fail"
df["Grade"] = df["Avg"].apply(assign_grades)
````
