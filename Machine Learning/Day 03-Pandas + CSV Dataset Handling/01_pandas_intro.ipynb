## Task 2
Create a Series with custom index:
```
Math = 85
Science = 90
English = 78
Computer = 95
```
Print each subject mark separately.


````python
sub=["Maths","Science","English","Computer"]
marks=pd.Series([85,90,78,92],index=sub)
marks
````
