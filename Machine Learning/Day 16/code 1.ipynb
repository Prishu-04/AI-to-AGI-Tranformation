```python
df_drop=df.dropna()
df_drop
```

```python
df_dro_col=df.drop("study_hours",axis=1)
df_dro_col
```

```python
df["study_hours"]=df["study_hours"].fillna(df["study_hours"].mean())
df
```

```python
df["attendance"]=df["attendance"].fillna(df["attendance"].median())
df
```

```python
df["study_method"]=df["study_method"].fillna(df["study_method"].mode()[0])
df
```

```python
df["study_method"]=df["study_method"].fillna("Unknown")
df
```

```python
df["prev_sc_miss"]=df["previous_score"].isnull().astype(int)
df
```

```python
df["previous_score"]=df["previous_score"].fillna(df["previous_score"].median())
df
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy="mean")

data = {
    "student_id": [101, 102, 103, 104, 105, 105, 106, 107],
    "study_hours": [5, 2, np.nan, 8, 6, 6, 1, 7],
    "attendance": [80, 45, 60, np.nan, 75, 75, 30, 85],
    "previous_score": [70, 40, 55, 90, np.nan, np.nan, 35, 88],
    "study_method": ["Online", "Offline", "Online", "Offline", np.nan, np.nan, "Offline", "Online"],
    "final_marks": [75, 42, 58, 92, 70, 70, 38, 90]
}

df1=pd.DataFrame(data)

df1[["study_hours", "attendance", "previous_score"]] = imputer.fit_transform(
    df1[["study_hours", "attendance", "previous_score"]]
)
df1
```

```python
cat_imputer = SimpleImputer(strategy="most_frequent")

df1[["study_method"]] = cat_imputer.fit_transform(df1[["study_method"]])
df1
```
