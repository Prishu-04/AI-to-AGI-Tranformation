```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "cgpa": [8.5, 6.2, 7.8, 5.9, 9.1, 8.0],
    "branch": ["CSE", "ECE", "CSE", "ME", "ECE", "CSE"],
    "internship": ["Yes", "No", "Yes", "No", "Yes", "No"],
    "skill_level": ["Advanced", "Beginner", "Intermediate", "Beginner", "Advanced", "Intermediate"],
    "placed": ["Yes", "No", "Yes", "No", "Yes", "Yes"]
}
df=pd.DataFrame(data)
df
```

```python
df["internship_en"]=df["internship"].map({
     "No":0,
     "Yes":1
})
df[["internship","internship_en"]]
```

```python
skill_order={
     "Beginner":0,
     "Intermediate":1,
     "Advanced":2
}
df["skill_level_en"]=df["skill_level"].map(skill_order)
df[["skill_level","skill_level_en"]]
```

```python
branch_encoded = pd.get_dummies(df["branch"], prefix="branch", dtype=int)
branch_encoded
```

```python
df_encoded = pd.concat([df, branch_encoded], axis=1)
df_encoded
```

```python
df_encoded = df_encoded.drop("branch", axis=1)
df_encoded
```

```python
df_encoded = pd.get_dummies(
    df,
    columns=["branch"],
    dtype=int
)

df_encoded
```

```python
df["placed_encoded"] = df["placed"].map({
    "No": 0,
    "Yes": 1
})

df[["placed", "placed_encoded"]]
```

```python
x= df[["cgpa", "internship_en", "skill_level_en"]]
y = df["placed_encoded"]
print(x)
print(y)
```

```python
from sklearn.preprocessing import OneHotEncoder

data = {
    "branch": ["CSE", "ECE", "CSE", "ME", "ECE", "CSE"]
}

df_branch = pd.DataFrame(data)

encoder = OneHotEncoder(sparse_output=False)

encoded_array = encoder.fit_transform(df_branch[["branch"]])

encoded_df = pd.DataFrame(
    encoded_array,
    columns=encoder.get_feature_names_out(["branch"])
)
encoded_df
```
