# Functions, Loops, Conditions & List Comprehension

##  Conditions in Python

### Syntax

```python
marks = 85
if marks >= 90:
    print("Excellent")
elif marks >= 75:
    print("Good")
else:
    print("Needs Improvement")
```

## AIML Use Case of Conditions

```python
accuracy = 0.92
if accuracy >= 0.90:
    print("Model is performing well")
else:
    print("Model needs improvement")
```

## For Loops

```python
marks = [80, 90, 75, 88]
for mark in marks:
    print(mark)
```


## 4. Using Loop to Calculate Average

```python
marks = [80, 90, 75, 88]
total = 0
for mark in marks:
    total = total + mark
average = total / len(marks)
print("Average:", average)
```

## While Loops

```python
count = 1
while count <= 5:
    print(count)
    count = count + 1
print("Finished counting")
```

## 6. Functions

```python
def greet():
    print("Welcome to AI/ML Roadmap")

greet()
```

## Function With Parameter

```python
def calculate_average(marks):
    average = sum(marks) / len(marks)
    return average
student_marks = [80, 90, 75, 88]
result = calculate_average(student_marks)
print("Average:", result)
```

## 8. AI/ML Use Case of Function

```python
def check_model_performance(accuracy):
    if accuracy >= 0.90:
        return "Excellent Model"
    elif accuracy >= 0.75:
        return "Good Model"
    else:
        return "Needs Improvement"
result = check_model_performance(0.86)
print(result)
```

## 9. List Comprehension

### Normal Method

```python
numbers = [1, 2, 3, 4, 5]
squares = []
for num in numbers:
    squares.append(num * num)
print(squares)
```

### List Comprehension Method

```python
numbers = [1, 2, 3, 4, 5]
squares = [num * num for num in numbers]
print(squares)
```

## 10. List comprehension with Condition

```python
marks = [45, 78, 90, 32, 88, 60]
passed_marks = [mark for mark in marks if mark >= 50]
print(passed_marks)
```

## 11. AI/ML Use Case of List comprehension

```python
values = [10, 20, 30, 40, 50]
normalized = [value / 100 for value in values]
print(normalized)
```
