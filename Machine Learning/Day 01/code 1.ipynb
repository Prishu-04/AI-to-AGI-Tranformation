### List of Lists

```python
dataset = [
    [5.1, 3.5, 1.4, 0.2],
    [4.9, 3.0, 1.4, 0.2],
    [6.2, 3.4, 5.4, 2.3]
]

print(dataset)
```

##### Accesing Each row

```python
for i in range(len(dataset)):
    print("Data Point", i+1, ":", dataset[i])
```

##### Accessing each value

```python
for i in range(len(dataset)):
    for j in range(len(dataset[i])):
        print("Value at position (",i+1,",",j+1,"):",dataset[i][j])
```
