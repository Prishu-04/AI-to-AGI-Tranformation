### Task 1
Create a list of marks:
```
marks = [80, 45, 90, 32, 70]
```
Print whether each student has passed or failed.
Passing marks: `50`

````python
marks=[80,45,90,32,70]
for mark in marks:
     if mark>50:
          print("pass")
````

### Task 2
Create a function:
```
calculate_average(marks)
```
It should return the average of marks.


````python
marks = [80, 45, 90, 32, 70]
total_marks = 0
for mark in marks:
     total_marks += mark
avg_marks = total_marks / len(marks)
print(avg_marks)
````

Task 3
Create a function:
```
check_grade(mark)
```
Rules:
```
90 and above = A
75 to 89 = B
50 to 74 = C
Below 50 = Fail

````python
def check_grade(mark):
     if mark >= 90 and mark <= 100:
          return "A"
     elif mark >= 75 and mark < 90:
          return "B"
     elif mark >= 50 and mark < 74:
          return "C"
     else:
          return "Fail"
     
marks = [80, 45, 90, 32, 70]
for mark in marks:
     grade = check_grade(mark)
     print(f"Mark: {mark}, Grade: {grade}")
````

### Task 4
Use list comprehension to create a list of squared numbers from:
```
numbers = [1, 2, 3, 4, 5]
```
Expected output:
```
[1, 4, 9, 16, 25]

````python
numbers = [1, 2, 3, 4, 5]
squared_numbers = [num*num for num in numbers]
print(squared_numbers)
````

### Task 5
Use list comprehension to filter only passed marks from:
```
marks = [80, 45, 90, 32, 70]
```
Expected output:
```
[80, 90, 70]

````python
marks = [80, 45, 90, 32, 70]
filtered_marks = [mark for mark in marks if mark > 50]
print(filtered_marks)
````
