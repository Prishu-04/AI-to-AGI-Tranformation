## Task 1
Create a program that divides two numbers.
Handle this error:
```
ZeroDivisionError

````python
try:
     num=2
     num1=0
     result=num/num1
     print(result)
except ZeroDivisionError:
     print("Cannot divide by zero")
````

## Task 2
Create a list:
```
marks = [80, 90, 75]
```
Try to access index `5`.
Handle this error:
```
IndexError

````python
marks=[80,90,75]
try:
     print(marks[5])
except IndexError:
     print("Index out of range")
````

````python
try:
     with open("day2_summary.txt","r") as file:
          content=file.read()
          print(content)
except FileNotFoundError:
    print("File not found.")
````
