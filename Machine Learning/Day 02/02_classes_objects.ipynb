# 02_classes_objects

## Create a class:
```
Product
```
Attributes:
```
name
price
rating
stock
```
Methods:
```
is_valid()
get_category()
to_dictionary()
show_info()
```
Rules:
```
Invalid:
price < 0 or rating < 0 or rating > 5 or stock < 0

Premium:
price >= 50000 and rating >= 4.5

Standard:
price >= 10000 and rating >= 3.5

Budget:
otherwise
```
Expected style:
```
Laptop => Premium
Phone => Standard
Keyboard => Budget
Fake Item => Invalid Product


````python
class product:
     def __init__(self, name, price,rate,stock):
          self.name = name
          self.price = price
          self.rate = rate
          self.stock = stock

     def is_valid(self):
          if self.price < 0 or self.rate < 0 or self.rate > 5 or self.stock < 0:
               return False
          return True
     
     def get_category(self):
          if self.price >= 50000 and self.rate >= 4.5:
               return "Premium"
          elif self.price >=10000 and self.rate >= 3.5:
               return "Standard"
          else:
               return "Budget"
     
     def to_dictionary(self):
          return {
               "name": self.name,
               "price": self.price,
               "rate": self.rate,
               "stock": self.stock
          }
     def show_info(self):
          print(self.name," => ", self.get_category())

product1 = product("Laptop", 60000, 4.7, 10)
product2 = product("Phone", 15000, 4.0, 20)
product3 = product("keyboard", 5000, 3.0, 30)
product4 = product("Fake Item", -2000, 4.0, 15)

products = [product1, product2, product3, product4]
for p in products:
     if p.is_valid():
          p.show_info()
     else:
          print(p.name, "=> Invalid Prduct")
````
