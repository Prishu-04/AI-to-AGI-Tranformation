## Instance Attribute

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

student1 = Student("Rahul", 85)
student2 = Student("Priya", 92)
print(student1.name)
print(student2.name)
```

## __init__() function

```python
class Student:
     def __init__(self, name, marks):
          self.name = name
          self.marks = marks

student1 = Student("Rahul", 85)
print(student1.name)
print(student1.marks)
```

## self

## Method

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def is_passed(self):
        if self.marks >= 40:
            return True
        return False

student1 = Student("Rahul", 85)
print(student1.is_passed())
```

## Different types of method usuage

```python
class Student:
    def __init__(self, name, marks, attendance):
        self.name = name
        self.marks = marks
        self.attendance = attendance

    def show_info(self):
        print(f"Name: {self.name}")
        print(f"Marks: {self.marks}")
        print(f"Attendance: {self.attendance}")

    def is_valid(self):
        if self.marks < 0 or self.marks > 100:
            return False

        if self.attendance < 0 or self.attendance > 100:
            return False

        return True

    def get_result(self):
        if self.marks >= 40 and self.attendance >= 75:
            return "Pass"
        return "Fail"

student1 = Student("Rahul", 85, 80)
student1.show_info()
print(student1.is_valid())
print(student1.get_result())
```

## Class Variable vs Instance Variable

### instance Variable

```python
class Student:
     def __init__(self, name, marks, attendance):
          self.name = name
          self.marks = marks
          self.attendance = attendance

s1= Student("Rahul", 85, 80)
s2= Student("Priya", 92, 90)
print(s1.name)
print(s2.name)
```

### Class Variable

```python
class Student:
     sch_name = "ABC School"
     def __init__(self, name, marks):
          self.name = name
          self.marks = marks
     
s1 = Student("Rahul", 85)
s2 = Student("Priya", 92)
print(s1.sch_name)
print(s2.sch_name)
```

## AI/ML Example: Dataset Record Class

```python
class StudentRecord:
    dataset_name = "student_performance_dataset"

    def __init__(self, name, marks, attendance, study_hours):
        self.name = name
        self.marks = marks
        self.attendance = attendance
        self.study_hours = study_hours

    def is_valid(self):
        if self.marks < 0 or self.marks > 100:
            return False

        if self.attendance < 0 or self.attendance > 100:
            return False

        if self.study_hours < 0:
            return False

        return True

    def predict_performance(self):
        if self.marks >= 80 and self.attendance >= 85 and self.study_hours >= 5:
            return "Excellent"
        elif self.marks >= 50 and self.attendance >= 75:
            return "Good"
        else:
            return "Needs Improvement"

    def to_dictionary(self):
        return {
            "name": self.name,
            "marks": self.marks,
            "attendance": self.attendance,
            "study_hours": self.study_hours,
            "prediction": self.predict_performance()
        }

student1 = StudentRecord("Rahul", 85, 80, 6)
print(student1.is_valid())
print(student1.predict_performance())
print(student1.to_dictionary())
```

## Batch processing with OOP

```python
students = [
    StudentRecord("Rahul", 85, 90, 5.5),
    StudentRecord("Aman", 45, 70, 3),
    StudentRecord("Priya", 72, 80, 4),
    StudentRecord("Sneha", 92, 95, 6),
    StudentRecord("Rohit", 120, 80, 4)
]

clean_data = []

for student in students:
    if student.is_valid():
        clean_data.append(student.to_dictionary())
        
for record in clean_data:
    print(record)
```
