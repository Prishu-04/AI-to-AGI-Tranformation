# day4_employee_analysis.py

```python
import pandas as pd
```

## Tasks:

1. Create employee DataFrame
2. Add level column:
   Senior: experience >= 5 and salary >= 70000
   Mid-Level: experience >= 2 and salary >= 40000
   Junior: otherwise
3. Filter Senior, Mid-Level, and Junior employees
4. Sort employees by salary descending
5. Calculate average salary
6. Calculate highest salary
7. Calculate average experience
8. Count levels
9. Count departments
10. Save as day4_employee_analysis_report.csv

## Create employee dataframe

```python
data = {
    "employee_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "name": ["Amit", "Riya", "Karan", "Neha", "Vikas", "Anjali", "Rohit", "Priya", "Rahul", "Sneha"],
    "salary": [25000, 45000, 80000, 120000, 35000, 70000, 30000, 95000, 55000, 40000],
    "experience": [1, 3, 6, 8, 2, 5, 1, 7, 4, 2],
    "rating": [3, 4, 5, 5, 2, 4, 3, 5, 4, 3],
    "department": ["HR", "IT", "AI", "AI", "Sales", "Data", "IT", "Data", "Web", "HR"]
}
df=pd.DataFrame(data)
print(df)
```

## Add level coloumn

```python
def get_level(row):
     if row["experience"]>=5 and row["salary"]>=70000:
          return "Senior"
     elif row["experience"]>=2 and row["salary"]>=40000:
          return "Mid-level"
     else:
          return "Junior"

df["level"]=df.apply(get_level,axis=1)
df
     
```

## 3. Filter Senior, Mid-Level, and Junior employees

```python
fl_level=df.loc[df["level"]=="Senior",["employee_id","name"]]
print(fl_level)
```

```python
fl1_level=df.loc[df["level"]=="Mid-Level",["employee_id","name"]]
print(fl1_level)
```

```python
fl2_level=df.loc[df["level"]=="Junior",["employee_id","name"]]
print(fl2_level)
```

## 4. Sort employees by salary descending

```python
sort=df.sort_values(by="salary",ascending=False)
sort
```

## 5. Calculate average salary


```python
avg=df["salary"].mean()
print(avg)
```

## 6. Calculate highest salary


```python
max_sal=df["salary"].max()
print(max_sal)
```

## 7. Calculate average experience


```python
avg_exp=df["experience"].mean()
print(avg_exp)
```

## 8. Count levels


```python
count_level=df["level"].sum()
print(count_level)
```

## 9. Count departments

```python
count_dept=df["department"].sum()
print(count_dept)
```

## 10 Save to csv file

```python
df.to_csv("day4_employee_analysis_report.csv", index=False)
```
