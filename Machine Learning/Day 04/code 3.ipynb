## Filtering Data

```python
high_math = df[df["Math"] > 80]
print(high_math)
```

## Filter Students with Attendance Above 90

```python
high_attendance = df[df["Attendance"] > 90]
print(high_attendance)
```

## Filter Students with Study Hours More Than 4

```python
hardworking_students = df[df["Study_Hours"] > 4]
print(hardworking_students)
```

## 13. Multiple Conditions

```python
result = df[(df["Math"] > 80) & (df["Attendance"] > 90)]
print(result)
```

```python
result = df[(df["Math"] > 90) | (df["Science"] > 90)]
print(result)
```

## Creating Total and Average Columns

```python
df["Total"] = df["Math"] + df["Science"] + df["English"]
df["Average"] = df["Total"] / 3
df["Average"] = df["Average"].round(2)
df
```

## Filter Passed Students
Pass condition:

Average >= 50

```python
passed_students = df[df["Average"] >= 50]
print(passed_students)
```

```python

16. Filter Failed Students
failed_students = df[df["Average"] < 50]

print(failed_students)
```

## Add Result Column

```python
df["Result"] = ["Pass" if avg >= 50 else "Fail" for avg in df["Average"]]
df
```

## Add Grade

```python
def assign_grade(avg):
    if avg >= 90:
        return "A+"
    elif avg >= 75:
        return "A"
    elif avg >= 60:
        return "B"
    elif avg >= 50:
        return "C"
    else:
        return "Fail"
df["Grade"] = df["Average"].apply(assign_grade)
df
```

## Filter using `loc[]`

```python
import pandas as pd

data = {
    "name": ["Rahul", "Aman", "Priya", "Sneha", "Karan", "Anjali", "Rohit"],
    "marks": [85, 72, 45, 91, 60, 88, 35],
    "attendance": [90, 80, 70, 95, 78, 92, 60],
    "study_hours": [5, 4, 3, 6, 2, 5, 1],
    "department": ["AI", "Data", "Web", "AI", "Data", "AI", "Web"]
}

df = pd.DataFrame(data)

print(df)
```

```python
result=df.loc[df["marks"]>=80, ["name","marks","department"]]
result
```

## Sort data by one column

### In Descending order

## Sort multiple Columns

```python
sortmul_df=df.sort_values(by=["department","marks"],ascending=[True,False])
sortmul_df
```

## Create result column using conditions


```python
def get_result(row):
    if row["marks"] >= 80 and row["attendance"] >= 85 and row["study_hours"] >= 5:
        return "Excellent"
    elif row["marks"] >= 50 and row["attendance"] >= 75:
        return "Good"
    else:
        return "Needs Improvement"


df["result"] = df.apply(get_result, axis=1)

print(df)
```
