## Dataset

```python
import pandas as pd

data = {
    "employee_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Amit", "Riya", "Karan", "Neha", "Vikas", "Anjali", "Rohit", "Priya"],
    "salary": [25000, 45000, 80000, 120000, 35000, 70000, 30000, 95000],
    "experience": [1, 3, 6, 8, 2, 5, 1, 7],
    "rating": [3, 4, 5, 5, 2, 4, 3, 5],
    "department": ["HR", "IT", "AI", "AI", "SALES", "DATA", "IT", "DATA"],
    "level": ["Junior", "Mid-Level", "Senior", "Senior", "Junior", "Senior", "Junior", "Senior"]
}

df = pd.DataFrame(data)
df
```

### 1. Average salary by department

```python
avg_sal=df.groupby("department")["salary"].mean()
avg_sal
```

### 2. Highest salary by department

```python
high_sal=df.groupby("department")["salary"].max()
high_sal
```

### 3. Lowest salary by department

```python
low_sal=df.groupby("department")["salary"].min()
low_sal
```

### 4. Employee count by department

```python
emp_count=df.groupby("department")["employee_id"].count()
emp_count
```

## 5. Average rating by level

```python
avg_rating=df.groupby("level")["rating"].mean()
avg_rating.round(3)
```

### 6. Average experience by level

```python
avg_exp=df.groupby("level")["experience"].mean()
avg_exp
```

### 7. Salary summary by department using agg()

```python
sal_sum=df.groupby("department")["salary"].agg(["mean","max","min","count"])
sal_sum
```

### 8. Group by department and level

```python
dep_lev=df.groupby(["department","level"])["employee_id"].count()
dep_lev
```

## 9. Convert grouped output to DataFrame using reset_index()

```python
report=dep_lev.reset_index()
report
```

### 10. Save summary to CSV

```python
df.to_csv("day6_employee_groupby_summary.csv",index=False)
```
