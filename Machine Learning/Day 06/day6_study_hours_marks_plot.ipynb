```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Dataset

```python
data = {
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Rahul", "Aman", "Priya", "Sneha", "Karan", "Anjali", "Rohit", "Neha"],
    "marks": [85, 72, 45, 91, 60, 88, 35, 67],
    "attendance": [90, 80, 70, 95, 78, 92, 60, 82],
    "study_hours": [5, 4, 3, 6, 2, 5, 1, 4],
    "subject": ["AI", "DATA", "AI", "AI", "DATA", "DATA", "WEB", "WEB"],
    "result": ["Excellent", "Good", "Needs Improvement", "Excellent", "Good", "Excellent", "Needs Improvement", "Good"]
}
```

```python
df=pd.DataFrame(data)
df
```

## Tasks:
1. Use student names on x-axis
2. Plot marks
3. Plot study_hours
4. Add title
5. Add x-axis label
6. Add y-axis label
7. Add legend
8. Add grid
9. Save as day6_study_hours_marks_plot.png

```python
plt.figure(figsize=(8,5))
plt.plot(df["name"],df["marks"],color="blue",marker="o",label="Marks")
plt.plot(df["name"],df["study_hours"],color="purple",marker="o",label="Study_Hours")
plt.title("Marks v/s Study hours")
plt.xlabel("Student Name")
plt.ylabel("Score")
plt.legend()
plt.grid(True)
plt.savefig("day6_study_hours_marks_plot.png")
plt.show()
```
