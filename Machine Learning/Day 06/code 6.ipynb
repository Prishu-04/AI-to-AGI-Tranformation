## Import Library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## DataSet

```python
data = {
    "employee_id": [
        101, 102, 103, 104, 105,
        106, 107, 108, 109, 110,
        111, 112, 113, 114, 115
    ],

    "name": [
        "Amit", "Riya", "Karan", "Neha", "Vikas",
        "Anjali", "Rohit", "Priya", "Rahul", "Sneha",
        "Arjun", "Meera", "Dev", "Pooja", "Varun"
    ],

    "salary": [
        28000, 45000, 80000, 120000, 36000,
        70000, 30000, 95000, 55000, 42000,
        65000, 105000, 50000, 75000, 33000
    ],

    "experience": [
        1, 3, 6, 9, 2,
        5, 1, 7, 4, 2,
        5, 8, 3, 6, 2
    ],

    "rating": [
        3, 4, 5, 5, 2,
        4, 3, 5, 4, 3,
        4, 5, 4, 4, 3
    ],

    "department": [
        "HR", "IT", "AI", "AI", "SALES",
        "DATA", "IT", "DATA", "WEB", "HR",
        "WEB", "AI", "DATA", "IT", "SALES"
    ],

    "level": [
        "Junior", "Mid-Level", "Senior", "Senior", "Junior",
        "Senior", "Junior", "Senior", "Mid-Level", "Mid-Level",
        "Mid-Level", "Senior", "Mid-Level", "Senior", "Junior"
    ]
}
```

```python
df=pd.DataFrame(data)
df
```

## Count Plot

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df,x="department")
plt.title("Employee Count Distribution")
plt.xlabel("Department")
plt.ylabel("Employee count")
plt.tight_layout()
plt.savefig("day6_department_countplot.png")
plt.show()
```

### COunt Plot using `hue`

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="department",
    hue="level"
)

plt.title("Employee Levels Across Departments")
plt.xlabel("Department")
plt.ylabel("Employee Count")
plt.tight_layout()

plt.savefig("day6_department_level_countplot.png")
plt.show()
```

## Histogram with `histoplot()`

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="salary",
    bins=6
)

plt.title("Employee Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Employee Count")
plt.tight_layout()

plt.savefig("day6_salary_histplot.png")
plt.show()
```

### Histogram with KDE

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="salary",
    bins=6,
    kde=True
)

plt.title("Employee Salary Distribution with KDE")
plt.xlabel("Salary")
plt.ylabel("Employee Count")
plt.tight_layout()

plt.savefig("day6_salary_kde_histplot.png")
plt.show()
```

## Box Plot

### Salary Distribution by Department

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="department",
    y="salary"
)

plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Salary")
plt.tight_layout()

plt.savefig("day6_department_salary_boxplot.png")
plt.show()
```

### Salary Distribution by Employee Level

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="level",
    y="salary"
)

plt.title("Salary Distribution by Employee Level")
plt.xlabel("Employee Level")
plt.ylabel("Salary")
plt.tight_layout()

plt.savefig("day6_level_salary_boxplot.png")
plt.show()
```

## Scatter plot

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="experience",
    y="salary"
)

plt.title("Experience vs Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.tight_layout()

plt.savefig("day6_experience_salary_scatterplot.png")
plt.show()
```

### using hue and size

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x="experience",
    y="salary",
    hue="level",
    size="rating"
)

plt.title("Experience vs Salary by Level and Rating")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.tight_layout()

plt.savefig("day6_advanced_scatterplot.png")
plt.show()
```

## Correaltion Heatmap

```python
numeric_columns = df[
    ["salary", "experience", "rating"]
]

correlation_matrix = numeric_columns.corr()

print(correlation_matrix)
```

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation_matrix,
    annot=True
)

plt.title("Employee Feature Correlation Heatmap")
plt.tight_layout()

plt.savefig("day6_employee_correlation_heatmap.png")
plt.show()
```
