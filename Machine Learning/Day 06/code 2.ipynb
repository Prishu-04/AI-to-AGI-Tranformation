## Dataset

```python
import pandas as pd

data = {
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Rahul", "Aman", "Priya", "Sneha", "Karan", "Anjali", "Rohit", "Neha"],
    "marks": [85, 72, 45, 91, 60, 88, 35, 67],
    "attendance": [90, 80, 70, 95, 78, 92, 60, 82],
    "study_hours": [5, 4, 3, 6, 2, 5, 1, 4],
    "subject": ["AI", "DATA", "AI", "AI", "DATA", "DATA", "WEB", "WEB"],
    "result": ["Excellent", "Good", "Needs Improvement", "Excellent", "Good", "Excellent", "Needs Improvement", "Good"]
}

df = pd.DataFrame(data)
df
```

## Group by one column

```python
avg_marks=df.groupby("subject")["marks"].mean()
avg_marks
```

## Count records in each group

```python
stu_sub=df.groupby("subject")["student_id"].count()
print(stu_sub)
```

```python
print(df["subject"].value_counts())
```

## Mulitple calculations using `agg()`

```python
subject_sum=df.groupby("subject")["marks"].agg(["mean","max","min","count"])
subject_sum
```

## Group multiple numeric columns

```python
sub_perf=df.groupby("subject")[["marks","attendance","study_hours"]].mean()
sub_perf
```

## Group by multiple columns

```python
sub_result=df.groupby(["subject","result"])["student_id"].count()
sub_result
```

## Convert groupby output to DataFrame

```python
sub_df=sub_result.reset_index()
sub_df
```
