## Import Librraies

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Dataset 

```python
data = {
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Rahul", "Aman", "Priya", "Sneha", "Karan", "Anjali", "Rohit", "Neha"],
    "marks": [85, 72, 45, 91, 60, 88, 35, 67],
    "attendance": [90, 80, 70, 95, 78, 92, 60, 82],
    "study_hours": [5, 4, 3, 6, 2, 5, 1, 4],
    "subject": ["AI", "DATA", "AI", "AI", "DATA", "DATA", "WEB", "WEB"],
    "result": ["Excellent", "Good", "Needs Improvement", "Excellent", "Good", "Excellent", "Needs Improvement", "Good"]
}
```

```python
df=pd.DataFrame(data)
df
```

## Matplotlib Plot - Marks by students

```python
plt.figure(figsize=(8,5))
plt.plot(df["name"],df["marks"])
plt.title("Student Marks")
plt.xlabel("Student name")
plt.ylabel("ylabel")
plt.grid(True)
plt.show()
```

## Improve the Plot with markers

```python
plt.figure(figsize=(8,5))
plt.plot(df["name"],df["marks"],marker="o")
plt.title("Student marks with markers")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

## Add Attendance Line

```python
plt.figure(figsize=(8,5))
plt.plot(df["name"], df["marks"], marker="o", label="Marks")
plt.plot(df["name"], df["attendance"], marker="s", label="Attendance")

plt.title("Marks vs Attendance")
plt.xlabel("Student Name")
plt.ylabel("Score")
plt.legend()
plt.grid(True)
plt.show()
```

## Save Plot as image 

```python
plt.figure(figsize=(8,5))
plt.plot(df["name"],df["marks"],marker="o",label="marks")
plt.title("Student Marks")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.legend()
plt.grid(True)
plt.savefig("day6_student_marks_plot.png")
plt.show()
```
