## Dataset

```python
import pandas as pd

data = {
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Rahul", "Aman", "Priya", "Sneha", "Karan", "Anjali", "Rohit", "Neha"],
    "marks": [85, 72, 45, 91, 60, 88, 35, 67],
    "attendance": [90, 80, 70, 95, 78, 92, 60, 82],
    "study_hours": [5, 4, 3, 6, 2, 5, 1, 4],
    "subject": ["AI", "DATA", "AI", "AI", "DATA", "DATA", "WEB", "WEB"],
    "result": ["Excellent", "Good", "Needs Improvement", "Excellent", "Good", "Excellent", "Needs Improvement", "Good"]
}

df = pd.DataFrame(data)

df
```

## Simple Aggregation

```python
print(df["marks"].mean())
```

```python
print(df["marks"].max())
```

```python
print(df["marks"].min())
```

```python
print(df["student_id"].count())
```

## Aggregation on multiple columns


```python
summ=df[["marks","attendance","study_hours"]].agg(["mean","max","min","count"])
summ
```

## Grouped aggregation with `agg()`

```python
sub_summ=df.groupby("subject").agg({
     "marks": ["mean", "max", "min"],
    "attendance": ["mean", "max", "min"],
    "study_hours": ["mean", "max", "min"],
    "student_id": "count"
})
sub_summ
```

## Named Aggregation

```python
subject_summary_clean = df.groupby("subject").agg(
    average_marks=("marks", "mean"),
    highest_marks=("marks", "max"),
    lowest_marks=("marks", "min"),
    average_attendance=("attendance", "mean"),
    average_study_hours=("study_hours", "mean"),
    student_count=("student_id", "count")
)

subject_summary_clean.round(2)
```

## Convert summary index into column

```python
sub_report=subject_summary_clean.reset_index()
sub_report
```

## Summary by result

```python
result_summary = df.groupby("result").agg(
    average_marks=("marks", "mean"),
    average_attendance=("attendance", "mean"),
    average_study_hours=("study_hours", "mean"),
    student_count=("student_id", "count")
).reset_index()
result_summary.round(3)
```

## Group by two columns

```python
subject_result_summary = df.groupby(["subject", "result"]).agg(
    average_marks=("marks", "mean"),
    student_count=("student_id", "count")
).reset_index()
subject_result_summary
```

## 10. Sort summary table

```python
subject_summary_clean = subject_summary_clean.sort_values(
    by="average_marks",
    ascending=False
)
subject_summary_clean.round(3)
```

## 11. Save summary tables

```python
subject_summary_clean.to_csv("day6_subject_summary_table.csv", index=False)
result_summary.to_csv("day6_result_summary_table.csv", index=False)
subject_result_summary.to_csv("day6_subject_result_summary_table.csv", index=False)
```
