```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Dataset

```python
data = {
    "employee_id": [
        101, 102, 103, 104, 105,
        106, 107, 108, 109, 110,
        111, 112, 113, 114, 115
    ],

    "name": [
        "Amit", "Riya", "Karan", "Neha", "Vikas",
        "Anjali", "Rohit", "Priya", "Rahul", "Sneha",
        "Arjun", "Meera", "Dev", "Pooja", "Varun"
    ],

    "salary": [
        28000, 45000, 80000, 120000, 36000,
        70000, 30000, 95000, 55000, 42000,
        65000, 105000, 50000, 75000, 33000
    ],

    "experience": [
        1, 3, 6, 9, 2,
        5, 1, 7, 4, 2,
        5, 8, 3, 6, 2
    ],

    "rating": [
        3, 4, 5, 5, 2,
        4, 3, 5, 4, 3,
        4, 5, 4, 4, 3
    ],

    "department": [
        "HR", "IT", "AI", "AI", "SALES",
        "DATA", "IT", "DATA", "WEB", "HR",
        "WEB", "AI", "DATA", "IT", "SALES"
    ],

    "level": [
        "Junior", "Mid-Level", "Senior", "Senior", "Junior",
        "Senior", "Junior", "Senior", "Mid-Level", "Mid-Level",
        "Mid-Level", "Senior", "Mid-Level", "Senior", "Junior"
    ]
}
```

```python
df=pd.DataFrame(data)
df
```

## Task:
1. Bar chart: employee name vs salary
2. Bar chart: employee count by department
3. Scatter plot: experience vs salary
4. Scatter plot: rating vs salary
5. Histogram: salary distribution
6. Histogram: experience distribution

```python
plt.figure(figsize=(9,5))
bar=plt.bar(df["name"],df["salary"])
plt.title("Employee name vs salary")
plt.xlabel("Employee Name")
plt.ylabel("Salary")
plt.xticks(rotation=90)
plt.grid(axis="y")
plt.bar_label(bar)
plt.tight_layout()
plt.savefig("day6_employeename_salary_bar_chart.png")
plt.show()
```

```python
emp_counts=df["department"].value_counts()
emp_counts
```

```python
plt.Figure(figsize=(8,5))
bars=plt.bar(emp_counts.index,emp_counts.values)
plt.title("Employee Count by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employee")
plt.bar_label(bars)
plt.grid(axis="y")
plt.tight_layout()
plt.savefig("day6_employe_counts_bar_chart.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["experience"], df["salary"])

plt.title("Experience v/s salary")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.grid(True)
plt.tight_layout()

plt.savefig("day6_experience_salary_scatter.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["rating"], df["salary"])

plt.title("Rating v/s salary")
plt.xlabel("Rating")
plt.ylabel("Salary")
plt.grid(True)
plt.tight_layout()

plt.savefig("day6_experience_salary_scatter.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["salary"], bins=5, edgecolor="black")

plt.title("Distribution of Salary")
plt.xlabel("Salary")
plt.ylabel("Number of Employee")
plt.grid(axis="y")
plt.tight_layout()

plt.savefig("day6_salary_histogram.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["experience"], bins=5, edgecolor="black")

plt.title("Distribution of Experience")
plt.xlabel("Employee")
plt.ylabel("Number of Employee")
plt.grid(axis="y")
plt.tight_layout()

plt.savefig("day6_experience_histogram.png")
plt.show()
```
