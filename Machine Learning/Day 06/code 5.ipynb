```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Datasets

```python
data = {
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": [
        "Rahul", "Aman", "Priya", "Sneha",
        "Karan", "Anjali", "Rohit", "Neha"
    ],
    "marks": [85, 72, 45, 91, 60, 88, 35, 67],
    "attendance": [90, 80, 70, 95, 78, 92, 60, 82],
    "study_hours": [5, 4, 3, 6, 2, 5, 1, 4],
    "subject": ["AI", "DATA", "AI", "AI", "DATA", "DATA", "WEB", "WEB"],
    "result": [
        "Excellent", "Good", "Needs Improvement", "Excellent",
        "Good", "Excellent", "Needs Improvement", "Good"
    ]
}
```

```python
df1=pd.DataFrame(data)
df1
```

## Line Plot

```python
plt.figure(figsize=(9, 5))

plt.plot(
    df1["name"],
    df1["marks"],
    marker="o",
    label="Marks"
)

plt.plot(
    df1["name"],
    df1["attendance"],
    marker="s",
    label="Attendance"
)

plt.title("Student Marks and Attendance")
plt.xlabel("Student Name")
plt.ylabel("Score")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("day6_line_plot.png")
plt.show()
```

## Bar Plots

```python
plt.figure(figsize=(9, 5))

bars = plt.bar(df1["name"], df1["marks"])

plt.title("Marks Scored by Each Student")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.xticks(rotation=45)
plt.grid(axis="y")
plt.bar_label(bars)
plt.tight_layout()

plt.savefig("day6_student_marks_bar_chart.png")
plt.show()
```

## Bar Chart for Categorical Counts

```python
result_counts=df1["result"].value_counts()
result_counts
```

```python
plt.figure(figsize=(8,5))
bars = plt.bar(result_counts.index, result_counts.values)

plt.title("Student Result Category Counts")
plt.xlabel("Result Category")
plt.ylabel("Number of Students")
plt.bar_label(bars)
plt.grid(axis="y")
plt.tight_layout()

plt.savefig("day6_result_counts_bar_chart.png")
plt.show()
```

## Scatter plots

```python
plt.figure(figsize=(8,5))
plt.scatter(df1["study_hours"], df1["marks"])

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.tight_layout()

plt.savefig("day6_study_hours_marks_scatter.png")
plt.show()
```

## Add student names to scatter plot

```python
plt.figure(figsize=(8, 5))

plt.scatter(df1["study_hours"], df1["marks"])

for index, row in df1.iterrows():
    plt.annotate(
        row["name"],
        (row["study_hours"], row["marks"])
    )

plt.title("Study Hours vs Marks with Student Names")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.tight_layout()

plt.savefig("day6_named_scatter_plot.png")
plt.show()
```

## Histogram

```python
plt.figure(figsize=(8, 5))

plt.hist(df1["marks"], bins=5, edgecolor="black")

plt.title("Distribution of Student Marks")
plt.xlabel("Marks Range")
plt.ylabel("Number of Students")
plt.grid(axis="y")
plt.tight_layout()

plt.savefig("day6_marks_histogram.png")
plt.show()
```

## Histograms for Attendance

```python
plt.figure(figsize=(8, 5))

plt.hist(df1["attendance"], bins=5, edgecolor="black")

plt.title("Distribution of Student Attendance")
plt.xlabel("Attendance Range")
plt.ylabel("Number of Students")
plt.grid(axis="y")
plt.tight_layout()

plt.savefig("day6_attendance_histogram.png")
plt.show()
```
