# 01_python_class_object

### Assignment


Create a class named:
```
Employee
```
It should have:
```
name
salary
experience
department
```
Add methods:
```
is_valid()
get_level()
show_info()
```
Rules:
```
Invalid:
salary < 0 or experience < 0
Senior:
experience >= 5 and salary >= 70000
Mid-Level:
experience >= 2 and salary >= 40000
Junior:
otherwise
```
Expected output:
```
Amit => Senior
Riya => Mid-Level
Karan => Junior
Invalid Employee
```

````python
class Employee:
    def __init__(self, name, salary, experience, department):
        self.name = name
        self.salary = salary
        self.experience = experience
        self.department = department

    def isvalid(self):
        return self.salary >= 0 and self.experience >= 0

    def get_level(self):
        if not self.isvalid():
            return "Invalid"
        elif self.salary >= 70000 and self.experience >= 5:
            return "Senior"
        elif self.salary >= 40000 and self.experience >= 2:
            return "Mid-Level"
        else:
            return "Junior"

n=int(input("Enter the number of employees: "))
for i in range(n):
     name = input("Enter employee name: ")
     salary = float(input("Enter employee salary: "))
     experience = int(input("Enter employee experience (in years): "))
     department = input("Enter employee department: ")
     
     emp = Employee(name, salary, experience, department)
     print(emp.name,"=>",emp.get_level())
````
