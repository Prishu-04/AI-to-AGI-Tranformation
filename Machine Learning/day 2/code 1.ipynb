## Constructor `__init__()`

```python
class student:
     def __init__(self,name,marks,attendance,study_hours):
          self.name = name
          self.marks = marks
          self.attendance = attendance
          self.study_hours = study_hours
```

```python
stu1=student("Pratyaksh",90,80,5)

print(stu1.name)
print(stu1.marks)
print(stu1.attendance)
print(stu1.study_hours)
```

## Add Method :

```python
class student:
     def __init__(self,name,marks,attendance,study_hours):
          self.name = name
          self.marks = marks
          self.attendance = attendance
          self.study_hours = study_hours

     def display_info(self):
          print("Name:",self.name)
          print("Marks:",self.marks)
          print("Attendance:",self.attendance)
          print("Study Hours:",self.study_hours)

stu1=student("Pratyaksh",90,80,5)
stu1.display_info()
```

## Add validation method

```python
class student:
     def __init__(self,name,marks,attendance,study_hours):
          self.name = name
          self.marks = marks
          self.attendance = attendance
          self.study_hours = study_hours
     
     def isvalis(self):
          if self.marks >= 40 and self.attendance >= 75:
               return True
          else:
               return False

stu1=student("Pratyaksh",90,80,5)
print(stu1.isvalis())
```

## Add Prediction method

```python
class student:
     def __init__(self,name,marks,attendance,study_hours):
          self.name = name
          self.marks = marks
          self.attendance = attendance
          self.study_hours = study_hours
     
     def predict(self):
          if self.marks >= 90 and self.attendance >= 90 and self.study_hours >= 5:
               return "Excellent"
          elif self.marks >= 75 and self.attendance >= 80 and self.study_hours >= 3:
               return "Good"
          elif self.marks >= 40 and self.attendance >= 75:
               return "Average"
          else:
               return "Needs Improvement"


stu1=student("Pratyaksh",90,80,5)
print(stu1.predict())
```

## Add Clear output

```python
class student:
     def __init__(self,name,marks,attendance,study_hours):
          self.name = name
          self.marks = marks
          self.attendance = attendance
          self.study_hours = study_hours

     def isvalis(self):
          if self.marks >= 40 and self.attendance >= 75:
               return True
          else:
               return False
     
     def predict(self):
          if self.marks >= 90 and self.attendance >= 90 and self.study_hours >= 5:
               return "Excellent"
          elif self.marks >= 75 and self.attendance >= 80 and self.study_hours >= 3:
               return "Good"
          elif self.marks >= 40 and self.attendance >= 75:
               return "Average"
          else:
               return "Needs Improvement"

n=int(input("Enter number of students: "))
for i in range(n):
     name = input("Enter student name: ")
     marks = int(input("Enter student marks: "))
     attendance = int(input("Enter student attendance: "))
     study_hours = int(input("Enter student study hours: "))
     stu=student(name,marks,attendance,study_hours)
     print(stu.isvalis())
     print(stu.predict())
```
